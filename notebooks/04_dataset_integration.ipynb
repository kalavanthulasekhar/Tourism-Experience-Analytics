```python
import pandas as pd
import numpy as np
```

```python
DATA_PATH = "../data/raw/"

transaction = pd.read_excel(DATA_PATH + "Transaction.xlsx")
user = pd.read_excel(DATA_PATH + "User.xlsx")
city = pd.read_excel(DATA_PATH + "City.xlsx")
country = pd.read_excel(DATA_PATH + "Country.xlsx")
region = pd.read_excel(DATA_PATH + "Region.xlsx")
continent = pd.read_excel(DATA_PATH + "Continent.xlsx")
item = pd.read_excel(DATA_PATH + "Item.xlsx")
type_df = pd.read_excel(DATA_PATH + "Type.xlsx")
mode = pd.read_excel(DATA_PATH + "Mode.xlsx")

print("All datasets loaded successfully.")
```

```python
def clean_columns(df):
    df = df.copy()
    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
    )
    return df


transaction = clean_columns(transaction)
user = clean_columns(user)
city = clean_columns(city)
country = clean_columns(country)
region = clean_columns(region)
continent = clean_columns(continent)
item = clean_columns(item)
type_df = clean_columns(type_df)
mode = clean_columns(mode)
```

```python
print("Transaction:", transaction.columns.tolist())
print("User:", user.columns.tolist())
print("City:", city.columns.tolist())
print("Country:", country.columns.tolist())
print("Region:", region.columns.tolist())
print("Continent:", continent.columns.tolist())
print("Item:", item.columns.tolist())
print("Type:", type_df.columns.tolist())
print("Mode:", mode.columns.tolist())
```

```python
print("Transaction VisitMode values:")
print(sorted(transaction["visitmode"].dropna().unique()))

print("\nMode table:")
print(mode)
```

```python
print("\nMode IDs:")
print(mode["visitmodeid"].unique())

print("\nMode names:")
print(mode["visitmode"].unique())
```

```python
master = transaction.merge(
    user,
    on="userid",
    how="left",
    validate="many_to_one"
)

print("Transaction rows:", len(transaction))
print("Master rows after User merge:", len(master))
```

```python
master = master.rename(columns={
    "continentid": "user_continentid",
    "regionid": "user_regionid",
    "countryid": "user_countryid",
    "cityid": "user_cityid"
})
```

```python
master.head()
```

```python
city_user = city.rename(columns={
    "cityid": "user_cityid",
    "cityname": "user_cityname",
    "countryid": "user_city_countryid"
})

master = master.merge(
    city_user,
    on="user_cityid",
    how="left",
    validate="many_to_one"
)

print("Rows after city merge:", len(master))
```

```python
country_user = country.rename(columns={
    "countryid": "user_countryid",
    "country": "user_country",
    "regionid": "user_country_regionid"
})

master = master.merge(
    country_user,
    on="user_countryid",
    how="left",
    validate="many_to_one"
)

print("Rows after country merge:", len(master))
```

```python
region_user = region.rename(columns={
    "regionid": "user_regionid",
    "region": "user_region",
    "continentid": "user_region_continentid"
})

master = master.merge(
    region_user,
    on="user_regionid",
    how="left",
    validate="many_to_one"
)

print("Rows after region merge:", len(master))
```

```python
continent_user = continent.rename(columns={
    "continentid": "user_continentid",
    "continent": "user_continent"
})

master = master.merge(
    continent_user,
    on="user_continentid",
    how="left",
    validate="many_to_one"
)

print("Rows after continent merge:", len(master))
```

```python
master = master.merge(
    item,
    on="attractionid",
    how="left",
    validate="many_to_one"
)

print("Rows after attraction merge:", len(master))
```

```python
master = master.merge(
    type_df,
    on="attractiontypeid",
    how="left",
    validate="many_to_one"
)

print("Rows after attraction type merge:", len(master))
```

```python
city_attraction = city.rename(columns={
    "cityid": "attractioncityid",
    "cityname": "attraction_cityname",
    "countryid": "attraction_countryid"
})

master = master.merge(
    city_attraction,
    on="attractioncityid",
    how="left",
    validate="many_to_one"
)

print("Rows after attraction city merge:", len(master))
```

```python
country_attraction = country.rename(columns={
    "countryid": "attraction_countryid",
    "country": "attraction_country",
    "regionid": "attraction_regionid"
})

master = master.merge(
    country_attraction,
    on="attraction_countryid",
    how="left",
    validate="many_to_one"
)

print("Rows after attraction country merge:", len(master))
```

```python
region_attraction = region.rename(columns={
    "regionid": "attraction_regionid",
    "region": "attraction_region",
    "continentid": "attraction_continentid"
})

master = master.merge(
    region_attraction,
    on="attraction_regionid",
    how="left",
    validate="many_to_one"
)

print("Rows after attraction region merge:", len(master))
```

```python
region_attraction = region.rename(columns={
    "regionid": "attraction_regionid",
    "region": "attraction_region",
    "continentid": "attraction_continentid"
})

master = master.merge(
    region_attraction,
    on="attraction_regionid",
    how="left",
    validate="many_to_one"
)

print("Rows after attraction region merge:", len(master))
```

```python
print("Final shape:", master.shape)

print("\nColumns:")
print(master.columns.tolist())

print("\nFirst 5 rows:")
display(master.head())
```

```python
missing = master.isnull().sum()

missing = missing[missing > 0].sort_values(ascending=False)

print("Columns containing missing values:")
print(missing)
```
