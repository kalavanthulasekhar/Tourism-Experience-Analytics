```python
cities_path = os.path.join(RAW_DATA_PATH, "Cities.xlsx")

cities_path = os.path.normpath(cities_path)

if not os.path.isfile(cities_path):
    alternative_path = os.path.join(RAW_DATA_PATH, "City.xlsx")

    if os.path.isfile(alternative_path):
        cities_path = os.path.normpath(alternative_path)
    else:
        raise FileNotFoundError(
            f"Neither Cities.xlsx nor City.xlsx was found at "
            f"{os.path.abspath(cities_path)}. "
            f"Available files: {os.listdir(RAW_DATA_PATH)}"
        )

cities = pd.read_excel(cities_path)

print("Cities dataset loaded successfully.")
```

DATA CLEANING & QUALITY ANALYSIS

```python
missing_city_name = cities[cities["CityName"].isnull()]

print("Rows with missing CityName:")
display(missing_city_name)
```

```python
missing_city_name_index = cities[cities["CityName"].isnull()].index

print("Missing CityName row index:", missing_city_name_index.tolist())

for index in missing_city_name_index:
    print("\nRecord:")
    print(cities.loc[index])
```

```python
for index in missing_city_name_index:
    start = max(0, index - 3)
    end = min(len(cities), index + 4)

    print(f"\nRecords around index {index}:")
    display(cities.iloc[start:end])
```

```python
for index in missing_city_name_index:
    city_id = cities.loc[index, "CityId"]
    country_id = cities.loc[index, "CountryId"]

    print(f"CityId   : {city_id}")
    print(f"CountryId: {country_id}")
```

```python
print(cities.dtypes)
```

```python
city_names = cities["CityName"].dropna().astype(str)

leading_spaces = city_names.str.startswith(" ").sum()
trailing_spaces = city_names.str.endswith(" ").sum()

print("Names with leading spaces :", leading_spaces)
print("Names with trailing spaces:", trailing_spaces)
```

```python
empty_city_names = (
    cities["CityName"]
    .fillna("")
    .astype(str)
    .str.strip()
    .eq("")
    .sum()
)

print("Empty CityName values:", empty_city_names)
```

```python
duplicate_city_names = (
    cities[cities["CityName"].notna()]
    .groupby("CityName")
    .size()
    .reset_index(name="Count")
    .sort_values("Count", ascending=False)
)

duplicate_city_names.head(20)
```

```python
repeated_names = duplicate_city_names[
    duplicate_city_names["Count"] > 1
]

print("Number of repeated city names:",
      len(repeated_names))

display(repeated_names.head(20))
```

```python
city_country_check = (
    cities[cities["CityName"].notna()]
    .groupby("CityName")["CountryId"]
    .nunique()
    .reset_index(name="UniqueCountryCount")
    .sort_values("UniqueCountryCount", ascending=False)
)

display(
    city_country_check[
        city_country_check["UniqueCountryCount"] > 1
    ].head(20)
)
```

```python
city_index = cities[cities["CityId"] == 6879].index[0]

display(
    cities.iloc[
        max(0, city_index - 5): city_index + 6
    ]
)
```

```python
country_151_cities = cities[
    cities["CountryId"] == 151
].sort_values("CityId")

print(f"Number of cities with CountryId 151: {len(country_151_cities)}")

display(country_151_cities)
```

```python
city_6879 = cities[cities["CityId"] == 6879]

print("CityId 6879 exists:", not city_6879.empty)
print("\nRecord:")
display(city_6879)
```

```python
expected_ids = set(range(cities["CityId"].min(),
                         cities["CityId"].max() + 1))

actual_ids = set(cities["CityId"])

missing_ids = expected_ids - actual_ids
extra_ids = actual_ids - expected_ids

print("Missing CityIds:", missing_ids)
print("Unexpected CityIds:", extra_ids)
```

```python
cities_cleaned = cities.copy()

print("Cleaning copy created.")
print("Original shape :", cities.shape)
print("Working shape  :", cities_cleaned.shape)
```

```python
cities_cleaned["CityName"] = (
    cities_cleaned["CityName"]
    .astype("string")
    .str.strip()
)

print("CityName whitespace cleaned.")
```

```python
empty_city_names = (
    cities_cleaned["CityName"]
    .fillna("")
    .str.strip()
    .eq("")
)

print("Empty CityName values:", empty_city_names.sum())
print("Missing CityName values:", cities_cleaned["CityName"].isna().sum())
```

```python
print("CityId missing:", cities_cleaned["CityId"].isna().sum())
print("CityId duplicates:", cities_cleaned["CityId"].duplicated().sum())

print("CountryId missing:", cities_cleaned["CountryId"].isna().sum())
```

```python
negative_city_ids = (cities_cleaned["CityId"] < 0).sum()
negative_country_ids = (cities_cleaned["CountryId"] < 0).sum()

print("Negative CityId values:", negative_city_ids)
print("Negative CountryId values:", negative_country_ids)
```

```python
print("CityId dtype   :", cities_cleaned["CityId"].dtype)
print("CountryId dtype:", cities_cleaned["CountryId"].dtype)
print("CityName dtype :", cities_cleaned["CityName"].dtype)
```

```python
print("Duplicate rows:", cities_cleaned.duplicated().sum())
print("Duplicate CityId:", cities_cleaned["CityId"].duplicated().sum())
```

```python
city_country_duplicates = (
    cities_cleaned[
        cities_cleaned["CityName"].notna()
    ]
    .duplicated(
        subset=["CityName", "CountryId"],
        keep=False
    )
)

print(
    "Duplicate CityName + CountryId records:",
    city_country_duplicates.sum()
)

display(
    cities_cleaned.loc[
        city_country_duplicates[city_country_duplicates].index
    ]
    .sort_values(["CityName", "CountryId"])
    .head(30)
)
```

```python
quality_report = pd.DataFrame({
    "Column": cities_cleaned.columns,
    "Data_Type": [
        str(cities_cleaned[col].dtype)
        for col in cities_cleaned.columns
    ],
    "Missing_Count": [
        cities_cleaned[col].isna().sum()
        for col in cities_cleaned.columns
    ],
    "Missing_Percentage": [
        round(cities_cleaned[col].isna().mean() * 100, 2)
        for col in cities_cleaned.columns
    ],
    "Unique_Values": [
        cities_cleaned[col].nunique(dropna=True)
        for col in cities_cleaned.columns
    ]
})

display(quality_report)
```

```python
print("=" * 60)
print("CITIES DATASET - CLEANING STATUS")
print("=" * 60)

print(f"Rows                  : {len(cities_cleaned):,}")
print(f"Columns               : {len(cities_cleaned.columns)}")
print(f"Duplicate rows        : {cities_cleaned.duplicated().sum()}")
print(f"Duplicate CityId      : {cities_cleaned['CityId'].duplicated().sum()}")
print(f"Missing CityId        : {cities_cleaned['CityId'].isna().sum()}")
print(f"Missing CityName      : {cities_cleaned['CityName'].isna().sum()}")
print(f"Missing CountryId     : {cities_cleaned['CountryId'].isna().sum()}")

print("\nCleaning actions completed:")
print("✓ CityName whitespace removed")
print("✓ CityId validated")
print("✓ CountryId validated")
print("✓ Duplicate rows checked")
print("✓ Duplicate CityId checked")
print("✓ Data types verified")

print("\n⚠ CityName for CityId 6879 is still missing.")
print("It will NOT be guessed or removed until related datasets are checked.")
```

```python
processed_path = "../data/processed/Cities_cleaned.csv"

cities_cleaned.to_csv(
    processed_path,
    index=False
)

print("Cleaned Cities dataset saved successfully:")
print(processed_path)
```
