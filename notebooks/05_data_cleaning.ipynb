```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)

master = pd.read_csv("../data/processed/tourism_master.csv")

print("Dataset shape:", master.shape)
display(master.head())
```

```python
print(master.dtypes)
```

```python
missing = master.isnull().sum()

missing_df = pd.DataFrame({
    "Missing_Count": missing,
    "Missing_Percentage": (missing / len(master) * 100).round(2)
})

missing_df = missing_df[missing_df["Missing_Count"] > 0]
missing_df = missing_df.sort_values("Missing_Count", ascending=False)

display(missing_df)
```

```python
for col in master.columns:
    if master[col].isnull().sum() > 0:
        print("\n" + "=" * 60)
        print("COLUMN:", col)
        print("=" * 60)
        display(master[master[col].isnull()].head(20))
```

```python
if "user_cityname" in master.columns:
    master["user_cityname"] = master["user_cityname"].fillna("Unknown City")

if "attraction_cityname" in master.columns:
    master["attraction_cityname"] = master["attraction_cityname"].fillna("Unknown City")
```

```python
user_geo_cols = [
    "user_cityid",
    "user_countryid",
    "user_regionid",
    "user_continentid"
]

available_cols = [c for c in user_geo_cols if c in master.columns]

display(
    master[master[available_cols].isnull().any(axis=1)][
        ["userid"] + available_cols
    ].drop_duplicates()
)
```

```python
geo_name_cols = [
    "user_cityname",
    "user_country",
    "user_region",
    "user_continent"
]

for col in geo_name_cols:
    if col in master.columns:
        master[col] = master[col].fillna("Unknown")
```

```python
print("Complete duplicate rows:", master.duplicated().sum())
```

```python
print("Duplicate TransactionIds:",
      master["transactionid"].duplicated().sum())
```

```python
print("Rating values:")
print(sorted(master["rating"].dropna().unique()))
```

```python
invalid_ratings = master[
    ~master["rating"].isin([1, 2, 3, 4, 5])
]

print("Invalid ratings:", len(invalid_ratings))

if len(invalid_ratings) > 0:
    display(invalid_ratings.head())
```

```python
print("Rating values:")
print(sorted(master["rating"].dropna().unique()))
```

```python
invalid_ratings = master[
    ~master["rating"].isin([1, 2, 3, 4, 5])
]

print("Invalid ratings:", len(invalid_ratings))

if len(invalid_ratings) > 0:
    display(invalid_ratings.head())
```

```python
print("Visit years:")
print(sorted(master["visityear"].dropna().unique()))

print("\nNumber of unique years:",
      master["visityear"].nunique())
```

```python
print(
    master[
        (master["visityear"] < 1900) |
        (master["visityear"] > 2100)
    ][["transactionid", "visityear"]].head()
)
```

```python
print("Visit months:")
print(sorted(master["visitmonth"].dropna().unique()))
```

```python
invalid_months = master[
    ~master["visitmonth"].isin(range(1, 13))
]

print("Invalid months:", len(invalid_months))
```

```python
print("VisitMode IDs:")
print(sorted(master["visitmode"].dropna().unique()))

print("\nVisitMode names:")
visitmode_names = {
    1: "Business",
    2: "Couples",
    3: "Family",
    4: "Friends",
    5: "Solo"
}

print(
    master["visitmode"]
    .map(visitmode_names)
    .value_counts()
)
```

```python
print("Unique attractions:",
      master["attractionid"].nunique())

print("\nMissing attraction IDs:",
      master["attractionid"].isnull().sum())

print("\nMissing attraction names:",
      master["attraction"].isnull().sum())
```

```python
text_columns = master.select_dtypes(include="object").columns

for col in text_columns:
    master[col] = master[col].str.strip()
```

```python
display(master[text_columns].head())
```

```python
categorical_columns = [
    "visitmode_name",
    "attraction",
    "attractiontype",
    "user_cityname",
    "user_country",
    "user_region",
    "user_continent",
    "attraction_cityname",
    "attraction_country",
    "attraction_region",
    "attraction_continent"
]

categorical_columns = [
    c for c in categorical_columns
    if c in master.columns
]

for col in categorical_columns:
    master[col] = master[col].astype("category")
```

```python
print("=" * 60)
print("FINAL CLEANING VALIDATION")
print("=" * 60)

print("Shape:", master.shape)

print("\nDuplicate rows:",
      master.duplicated().sum())

print("\nDuplicate TransactionIds:",
      master["transactionid"].duplicated().sum())

print("\nMissing values:")
print(
    master.isnull().sum()[
        master.isnull().sum() > 0
    ]
)

print("\nRating values:")
print(sorted(master["rating"].dropna().unique()))

print("\nVisit modes:")
print(
    master["visitmode"]
    .map(visitmode_names)
    .value_counts()
)

print("\nAttractions:",
      master["attractionid"].nunique())
```

```python
output_path = "../data/processed/tourism_master_cleaned.csv"

master.to_csv(
    output_path,
    index=False
)

print(f"✅ Cleaned dataset saved to: {output_path}")
```
